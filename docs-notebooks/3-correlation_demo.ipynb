## Plot examples

### MCC figure

```python
#
# NOTE : You can have better looking figures by using the Holoviz interactive map below and clicking on the save button
#

%matplotlib inline
TH_SWA = 0.15
TH_YA = -0.5
vis_corr.plot_mcc_map(corr_ds, th_swa=TH_SWA, th_ya=TH_YA, show=True)
```

### Maximum MCC figure

```python
max_mmc_map_file = f"{config.corr_config.CORR_OUTPUT_DIR}/max_mcc_map.png"

if not os.path.exists(max_mmc_map_file):
	vis_corr.plot_max_mcc_map(corr_ds, save=True, show=False)
	print(f"Max MCC map saved to {max_mmc_map_file}")

display(Image(filename=max_mmc_map_file))
```
